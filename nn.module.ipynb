```python
import torch

import torch.nn as nn

# Trainig data
x = torch.tensor([[1.0],[2.0],[3.0],[4.0]])
y = torch.tensor([[2.0],[4.0],[6.0],[8.0]])

# Create Model
class MyModel (nn.Module):
    
    def __init__(self):
        super().__init__()

        self.linear = nn.Linear(1,1)
        
    def forward(self,x):
        return self.linear(x)
    
    
model = MyModel()


# Loss function
loss_function = nn.MSELoss()

# Optimizer
optimizer = torch.optim.SGD(model.parameters(), lr = 0.01)

# Trainig
for epoch in range(500):
    
    # prediction
    y_pred = model(x)
    
    # Calculate loss
    loss = loss_function(y_pred, y)
    
    # Calculate gradients
    optimizer.zero_grad()
    loss.backward()
    
    # Update Parameters
    optimizer.step()
    
    if epoch % 100 == 0:
        print(f"Epoch {epoch}: Loss = {loss.item():.4f}")
    
# Test the model
with torch.no_grad():
        
    new_x = torch.tensor([[5.0]])
        
    prediction = model(new_x)
        
print("Prediction :", prediction.item())    
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

class MyModel (nn.Module):
    
    def __init__(self):    #Ye model ke parts/layers banane ki jagah hai.
        super().__init__() #nn.Module ke basic setup ko initialize karta hai.
        
        self.Linear = nn.Linear(1,1)  #Yahan hum model ke andar ek Linear Layer bana rahe hain.
        
    def forward(self, x):  #"Input x model ke andar kis route se jayega?
        return self.Linear(x)
    
x = torch.tensor([[1.0],[2.0],[3.0],[4.0]])
y = torch.tensor([[2.0],[4.0],[6.0],[8.0]])
    
model = MyModel()
    
loss_function = nn.MSELoss()
    
optimizer = optim.Adam(model.parameters(), lr = 0.1)
    
for epoch in range(10):
        
    optimizer.zero_grad()
    y_pred = model(x)
    loss = loss_function(y_pred,y)
    loss.backward()
        
    optimizer.step()
    
    print(f"Epoch {epoch +1}, Loss = {loss.item():.4f}")        
    
torch.save(model. state_dict(), "model.pth")
print("Model saved Successfully!")

# Load Model

loaded_model = MyModel()

loaded_model.load_state_dict(
    torch.load("model.pth")
)

loaded_model.eval()   # "Ab model ko training nahi karni, sirf prediction/inference karna hai."

with torch.no_grad():
    new_x = torch.tensor([[6.0]])
    prediction = loaded_model(new_x)

print("Prediction:", prediction.item())
```
