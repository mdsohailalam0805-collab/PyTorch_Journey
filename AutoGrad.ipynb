```python
# small project

import torch

# step1, input data x

x = torch.tensor([1.,2.,3.,4.,5])

# step2, y

y = torch.tensor([3.,6.,9.,12.,15]) 

# step3 wrong guesses from w and b

w = torch.tensor(0.0, requires_grad=True)

b = torch.tensor(0.0, requires_grad=True)

# step 4

y_pred = w * x + b

loss = ((y_pred - y)**2).mean()

print("Predicted Value :", y_pred)

print("Actual Value :", y)

print("Loss :", loss)

loss.backward()

print("Gradient of w:", w.grad)

print("Gradient of b:", b.grad)
```

```python
import torch

x = torch.tensor([1., 2., 3., 4., 5.])
y = torch.tensor([3., 6., 9., 12., 15.])

w = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

optimizer = torch.optim.SGD([w, b], lr=0.01)

for epoch in range(1000):
    optimizer.zero_grad()
    
    y_pred = w * x + b
    
    loss = ((y_pred - y)**2).mean()
    
    loss.backward()
    
    optimizer.step()
    
    if epoch % 100 == 0:  # 100 is not a special number. It is just a printing interval.
        
        print(epoch, loss.item())
```

```python
x = torch.tensor([1.,2.,3.,4])
y = torch.tensor([2.,4.,6.,8])   # correct numbers

w = torch.tensor(0.0 , requires_grad = True)

lr = 0.1

epochs = 10

for epoch in range(epochs):
    
    # step 1 prediction
    
    y_pred = w * x
    
    # step 2 loss
    
    loss = ((y_pred - y)**2).mean()
    loss.backward()
    
    with torch.no_grad():
        w -= lr * w.grad
        
    # print(epochs , loss.item())
    
    print(f"Epoch {epoch + 1}: w={w.item():.4f}, loss= {loss.item():.4f}")
    

```
